##Task 2 – Add New Customer Records**

```python
from datetime import datetime


saas_data = {
    "customer_id": [101, 102, 103, 104, 105, 106, 107],
    "name": [
        "Rahul Sharma", "Meena Nair", None, "A. Kumar",
        "David", "Priya", "Rahul Sharma"
    ],
    "signup_date": [
        "2023-01-10", "2023/02/15", "03-01-2023",
        "15-04-2023", "2023-05-10", "20230520",
        "2023-01-10"
    ],
    "plan": [
        "Pro", "Basic", "Pro", "Enterprise",
        "Pro", "", "Pro"
    ],
    "monthly_fee": [
        "49.99", "19.99", "forty-nine", "99.99",
        "49.99", "49.99", "49.99"
    ],
    "usage_hours": [
        "120", "45", "88", None,
        "NULL", "200", "120"
    ]
}


print("Initial Data")
print()
print(saas_data)

totalNum = int(input("How many customers do you want to add? "))


if totalNum <= 0:

    print("Please enter a valid number of customers")

else:

    for i in range(totalNum):


        name = input("Enter customer name: ")


        date_string = input(
            "Enter signup date "
            "(YYYY-MM-DD / YYYY/MM/DD / DD-MM-YYYY / DD/MM/YYYY): "
        )


        date_formats = [
            "%Y-%m-%d",
            "%Y/%m/%d",
            "%d-%m-%Y",
            "%d/%m/%Y",
            "%m-%d-%Y",
            "%m/%d/%Y",
            "%Y%m%d"
        ]

        formatted_date = None

        for date_format in date_formats:

            try:
                formatted_date = datetime.strptime(
                    date_string,
                    date_format
                )
                break

            except ValueError:
                continue


        if formatted_date is None:

            print("Invalid date format.")
            print("Please enter a valid date.")

            continue


        formatted_date_new = formatted_date.strftime("%Y-%m-%d")

        print("Formatted Date:", formatted_date_new)


        plan = input("Enter Plan: ")


        monthsFees = input("Enter monthly fee: ")


        usageHours = input("Enter usage hours: ")


        generateCustId = max(saas_data["customer_id"]) + 1

        print("Generated Customer ID:", generateCustId)


        saas_data["customer_id"].append(generateCustId)
        saas_data["name"].append(name)
        saas_data["signup_date"].append(formatted_date_new)
        saas_data["plan"].append(plan)
        saas_data["monthly_fee"].append(monthsFees)
        saas_data["usage_hours"].append(usageHours)


print()
print("Please check the total list details")
print(saas_data)
```

##Task 3 – Create SaaS Customer Class (OOP)

```python
class Customer:
  def __init__(self,Customer_id,Name,Signup_date,Plan,Monthly_fee,usage_hours):
    self.Customer_id=Customer_id
    self.Name=Name
    self.Signup_date=Signup_date
    self.Plan=Plan
    self.Monthly_fee=float(Monthly_fee)
    self.usage_hours=usage_hours

  def validate_data(self):
    print("INside Validate Data")

    if self.Name =="":
      self.Name="Unknown Customer."
    else:
      self.Name=Name


    Plan = self.Plan.strip()
    Plan = " ".join(Plan.split())
    self.Plan = Plan.lower()


    if self.usage_hours is not None and self.usage_hours != "":
      self.usage_hours=int(self.usage_hours)
      if self.usage_hours < 0:
        print("Invalid value. Number cannot be negative.")
      else:
        try:
          print("Valid usage_hours:", self.usage_hours)

        except (ValueError, TypeError):
          self.usage_hours = 0
    else:
      self.usage_hours = 0

  def display_customer(self):
    print(f'')
    print("Customer Details")
    print(f'')
    print("Customer ID:::",{self.Customer_id})
    print("Customer Name:::",{self.Name})
    print("Signup_date:::",{self.Signup_date})
    print("Plan:::",{self.Plan})
    print("Monthly_fee:::",{self.Monthly_fee})
    print("usage_hours:::",{self.usage_hours})

customer1=Customer('109','','2025-01-30','    Pro      NEW   ','34','2')

customer1.validate_data()
customer1.display_customer()
```

##Task 4 – Data Cleaning Pipeline

```python
from datetime import datetime


class Customer:

    saas_data = {
        "customer_id": [101, 102, 103, 104, 105, 106, 107, 108],

        "name": [
            "Rahul Sharma", "Meena Nair", None, "A. Kumar",
            "David", "Priya", "Rahul Sharma", "Rahul"
        ],

        "signup_date": [
            "2023-01-10", "2023/02/15", "03-01-2023",
            "15-04-2023", "2023-05-10", "20230520",
            "2023-01-10", "2023-01-10"
        ],

        "plan": [
            "Pro", "Basic", "Pro", "Enterprise",
            "Pro", "", "Pro", "Pro"
        ],

        "monthly_fee": [
            "49.99", "19.99", "forty-nine", "99.99",
            "49.99", "49.99", "49.99", "49.99"
        ],

        "usage_hours": [
            "", "45", "88", None,
            "NULL", "200", "120", "120"
        ]
    }


    def __init__(self, saas_data):

        self.saas_data = saas_data

        print("Initial Data")
        print(self.saas_data)

    def Standardize_date_formats(self):

        signup_date_new = []

        date_formats = [
            "%Y-%m-%d",
            "%Y/%m/%d",
            "%m-%d-%Y",
            "%d-%m-%Y",
            "%Y%m%d"
        ]

        for date_string in self.saas_data["signup_date"]:

            formatted_date = None

            for date_format in date_formats:

                try:
                    formatted_date = datetime.strptime(
                        date_string,
                        date_format
                    )
                    break

                except (ValueError, TypeError):
                    continue

            if formatted_date:

                formatted_date_new = formatted_date.strftime(
                    "%Y-%m-%d"
                )

                signup_date_new.append(formatted_date_new)

            else:

                signup_date_new.append("Invalid Date")

        self.saas_data["signup_date"] = signup_date_new

        return self.saas_data


    def handle_missing_values(self):


        for i in range(len(self.saas_data["customer_id"])):

            value = self.saas_data["name"][i]

            if value is None or str(value).strip().lower() in [
                "",
                "null",
                "none"
            ]:

                self.saas_data["name"][i] = "Unknown Customer"

        for i in range(len(self.saas_data["customer_id"])):

            value = self.saas_data["usage_hours"][i]

            if value is None or str(value).strip().lower() in [
                "",
                "null",
                "none"
            ]:

                self.saas_data["usage_hours"][i] = 0

            else:

                try:
                    self.saas_data["usage_hours"][i] = int(value)

                except (ValueError, TypeError):

                    self.saas_data["usage_hours"][i] = 0


        for i in range(len(self.saas_data["monthly_fee"])):

            value = self.saas_data["monthly_fee"][i]

            try:

                self.saas_data["monthly_fee"][i] = float(value)

            except (ValueError, TypeError):

                print("Invalid monthly fee:", value)

                self.saas_data["monthly_fee"][i] = None

        return self.saas_data


    def remove_duplicate_records(self):

        unique_names = set()
        unique_records = []

        for i in range(len(self.saas_data["customer_id"])):

            name = self.saas_data["name"][i]

            if name not in unique_names:

                unique_names.add(name)
                unique_records.append(i)

        for key in self.saas_data:

            self.saas_data[key] = [
                self.saas_data[key][i]
                for i in unique_records
            ]

        print("\nDuplicate customer names removed.")

        return self.saas_data


customer1 = Customer(Customer.saas_data)

date_saas_data = customer1.Standardize_date_formats()

print("\nDate values are formatted:")
print(date_saas_data)


dupli_saas_data = customer1.remove_duplicate_records()

print("\nAfter removing duplicate customer names:")
print(dupli_saas_data)



saas_data_missing = customer1.handle_missing_values()

print("\nAfter handling missing values:")
print(saas_data_missing)
```
